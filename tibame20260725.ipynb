```python
# oxoxox
# xoxoxo
# oxoxox
# xoxoxo

# 一行
# oxoxox四行
# 輪流
col, row = 6, 4
symbols = ["o", "x"]


for i in range(row):
    start = i % len(symbols)
    line = ""
    for j in range(col):
        pos = (start + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```

```python
# 初始化一個雙層的list
# 錯的(要確認一個東西創了幾次)
a = [0] * 2
l = [a] * 3
l[0][0] = 99
print(l)

# 2. 對的
l = []
for i in range(3):
    a = [0] * 2
    l.append(a)
l[0][0] = 99
print(l)

# 3. 2的簡化
l = [[0] * 2 for i in range(3)]
l[0][0] = 88
print(l)
```

```python
# [0, 1, 2, 3, 4]
# [9, 8, 7, 6, 5]
# [10, 11, 12, 13, 14]
# [19, 18, 17, 16, 15]

# 準備雙層的list
col, row = 5, 10
start = 19
result = [[0] * col for i in range(row)]

# 準備一個雙層的迴圈來做設定
# 一列 -> 都是順的 -> 蛇狀的
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            result[i][j] = col * i + j + start
        else:
            result[i][j] = col * i + (col - 1) - j + start

print(result)
```

```python
# 群集型態:
# list(有序): 同類型的一群 []
# dict(無序): 拿來表示一個複雜東西 {key:value}
# set(無序)(特化的list): 裡面元素永不重複的一個集合 {}
# tuple

s = {20, 30, 15, 40, 30}
print(s)

# 交集: & 列出我們共同的
# 聯集: | 把所有都列出
# 減集: - 我的 - 我們共同的

s1 = {20, 30, 10}
s2 = {40, 30}

print(s1 & s2)
print(s1 | s2)
print(s1 - s2)

# 迴圈
for s in s1:
    print("!", s)

# 專屬功能: add
s1.add(80)
print(s1)
```

```python
# 樂透彩券: 不重複的六個號碼(1-49)
import random

# 如果你用list
lottery = []
while True:
    n = random.randint(1, 49)
    # print(n)
    if not n in lottery:
        lottery.append(n)
    if len(lottery) == 6:
        break
# print(lottery)

import random
# 如果你用set
lottery = set()
while len(lottery) < 6:
    n = random.randint(1, 49)
    lottery.add(n)

print(lottery)
prize = {2, 5, 19, 20, 39, 15}
print(lottery & prize)
```

```python
import pygame as pg
import random


#pygame初始化
pg.init()

# --- 字型 ---
def get_font(size):
    for f in ['microsoftjhenghei', 'simhei', 'stheitirelight']:
        if f in pg.font.get_fonts():
            return pg.font.SysFont(f, size)
    return pg.font.SysFont(None, size)


# GAME常數
FONT_UI = get_font(48)
col, row = 8, 8
inter = 80
width, height = (col + 2) * inter, (row + 2) * inter


# GAME UI
# 產生視窗
screen = pg.display.set_mode([width, height])
# 設定遊戲標題
pg.display.set_caption("翻牌")
# 準備第一個圖層
bg = pg.Surface(screen.get_size())
# 把畫布填滿某個顏色
bg.fill([199, 167, 82])
# 你要把圖層放到上一層
screen.blit(bg, [0, 0])
# 對畫面進行更新(才會真的秀出來)
pg.display.update()


# 建立一個永不結束的迴圈(遊戲才不會結束)
running = True
while running:
    # 收取你的遊戲任何事件(滑鼠點擊/鍵盤按鈕...)
    for event in pg.event.get():
        # 偵測滑鼠點擊以後放掉的動作
        if event.type == pg.MOUSEBUTTONUP:
            pass
        # 如果收到的事件是按x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False

pg.quit()
```
