```python
# 1. 雙層迴圈: 寫完內層以後, 把那個迴圈當作一行程式碼
# start 1
# hello
# hello
# hello
# start 2

for i in range(2):
    print("start", i+1)
    for j in range(3):
        print("hello")
```

```python
def func():
    for j in range(3):
        print("hello")

for i in range(2):
    print("start", i+1)
    func()
```

```python
# oxoxox
# xoxoxo
# oxoxox
# xoxoxo

# 一行
# oxoxox四行
# 輪流
col, row = 6, 4
symbols = ["o", "x"]


for i in range(row):
    start = i % len(symbols)
    line = ""
    for j in range(col):
        pos = (start + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```

```python
# 多層list: 很多人在想兩層的list,會用表格想(X)
l = [20, [[8, 18, 28], 19, 29], 40]
l[1][0][2]
```

```python
# 雙層你要超熟
# l[i][j] 格式
l = [
    [9, 19, 29],
    [8, 18, 28],
    [7, 17, 27]
]
l[1][2]
l[2][1] = 99
l
```

```python
# 初始化一層的list
# 1. *
l = [0] * 3
print(l)
# 2. 如果你裡面東西要不一樣
l = []
for i in range(4):
    l.append(i)
print(l)
# 3. 是2的簡化
l = [i+1 for i in range(4)]
print(l)
```

```python
# 初始化一個雙層的list
# 錯的(要確認一個東西創了幾次)
a = [0] * 2
l = [a] * 3
l[0][0] = 99
print(l)
# 2. 對的
l = []
for i in range(3):
    a = [0] * 2
    l.append(a)
l[0][0] = 99
print(l)
```
