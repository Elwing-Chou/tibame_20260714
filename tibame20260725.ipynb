```python
# 1. 雙層迴圈: 寫完內層以後, 把那個迴圈當作一行程式碼
# start 1
# hello
# hello
# hello
# start 2

for i in range(2):
    print("start", i+1)
    for j in range(3):
        print("hello")
```

```python
def func():
    for j in range(3):
        print("hello")

for i in range(2):
    print("start", i+1)
    func()
```

```python
# oxoxox
# xoxoxo
# oxoxox
# xoxoxo

# 一行
# oxoxox四行
# 輪流
col, row = 6, 4
symbols = ["o", "x"]


for i in range(row):
    start = i % len(symbols)
    line = ""
    for j in range(col):
        pos = (start + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```

```python
# 多層list: 很多人在想兩層的list,會用表格想(X)
l = [20, [[8, 18, 28], 19, 29], 40]
l[1][0][2]
```

```python
# 初始化一層的list
# 1. *
l = [0] * 3
print(l)
# 2. 如果你裡面東西要不一樣
l = []
for i in range(4):
    l.append(i)
print(l)
# 3. 是2的簡化
l = [i+1 for i in range(4)]
print(l)
```

```python
# 初始化一個雙層的list
# 錯的(要確認一個東西創了幾次)
a = [0] * 2
l = [a] * 3
l[0][0] = 99
print(l)

# 2. 對的
l = []
for i in range(3):
    a = [0] * 2
    l.append(a)
l[0][0] = 99
print(l)

# 3. 2的簡化
l = [[0] * 2 for i in range(3)]
l[0][0] = 88
print(l)
```

```python
# [0, 1, 2, 3, 4]
# [9, 8, 7, 6, 5]
# [10, 11, 12, 13, 14]
# [19, 18, 17, 16, 15]

# 準備雙層的list
col, row = 5, 10
start = 19
result = [[0] * col for i in range(row)]

# 準備一個雙層的迴圈來做設定
# 一列 -> 都是順的 -> 蛇狀的
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            result[i][j] = col * i + j + start
        else:
            result[i][j] = col * i + (col - 1) - j + start

print(result)
```
