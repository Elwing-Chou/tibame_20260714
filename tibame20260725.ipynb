```python
# 1. 雙層迴圈: 寫完內層以後, 把那個迴圈當作一行程式碼
# start 1
# hello
# hello
# hello
# start 2

for i in range(2):
    print("start", i+1)
    for j in range(3):
        print("hello")
```

```python
def func():
    for j in range(3):
        print("hello")

for i in range(2):
    print("start", i+1)
    func()
```

```python
# oxoxox
# xoxoxo
# oxoxox
# xoxoxo

# 一行
# oxoxox四行
# 輪流
col, row = 6, 4
symbols = ["o", "x"]

for i in range(row):
    start = i % len(symbols)
    line = ""
    for j in range(col):
        pos = (start + j) % len(symbols)
        line = line + symbols[pos]
    print(line)
```
