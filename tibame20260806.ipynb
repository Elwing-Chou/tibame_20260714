```python
import json
import urllib.request as req

url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn=26850&geo=TW%2CHK"
f = req.urlopen(url)
# list + dict: JSON格式
content = f.read()
print(type(content))
content = json.loads(content)
print(type(content))

danmus = content["data"]["danmu"]
for d in danmus:
    uid = d["userid"]
    text = d["text"]
    print(uid, text)
```
