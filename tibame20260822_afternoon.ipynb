````python
import urllib.request as req
import bs4 as bs

url = "https://tabelog.com/tw/tokyo/rstLst/sweets/?SrtT=rt"
resp = req.urlopen(url)
content = resp.read()
html = bs.BeautifulSoup(content)
# html.find("a", {"class":"list-rst__rst-name-target"})
rs = html.find_all("div", {"class":"list-rst__body"})
for r in rs:
    # 拿取區塊
    name = r.find("a", {"class":"list-rst__rst-name-target"})

    # 區塊萃取
    name_href = name["href"]
    name_text = name.get_text()

    # 印出來
    print(name_text)
    print(name_href)
````
